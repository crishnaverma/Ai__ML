```python
import pandas as pd
import seaborn as sns
from sklearn.preprocessing import StandardScaler
```

```python
data = pd.read_csv("thyroid_dataset.csv")
data
```

```python
X = data.drop("Outlier_label",axis = 1)
y = data["Outlier_label"]
```

```python
scaler = StandardScaler()
X_scale = scaler.fit_transform(X)
```

```python
from sklearn.ensemble import IsolationForest
If = IsolationForest(n_estimators = 200, random_state = 42)
label = If.fit_predict(X_scale)
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components = 2)
X_pca = pca.fit_transform(X_scale)
```

```python
sns.scatterplot(x = X_pca[:,0], y = X_pca[:,1], c = label)
```

```python
import numpy as np
print("no_outlier = ",np.sum(label == -1))
print("no_normal_data = ",np.sum(label == 1))
```
