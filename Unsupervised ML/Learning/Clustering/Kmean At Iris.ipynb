```python
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
import seaborn as sns
from sklearn.preprocessing import StandardScaler
```

```python
data = load_iris()
X = data.data
y = data.target
```

```python
standard = StandardScaler()
X_scaled = standard.fit_transform(X)
```

```python
from sklearn.metrics import silhouette_score
```

Error Case

Although the Iris dataset contains three known species, the silhouette analysis may select K=2 because Setosa is strongly separated while Versicolor and Virginica have overlapping feature distributions

```python
ss = []
for i in range(2,10):
    kmean = KMeans(n_clusters = i)
    label = kmean.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, label)
    ss.append(score)
```

```python
sns.lineplot(x=range(2,10), y = ss, marker = "o")
```

Taking K = 3 based on data knowledge

```python
kmean = KMeans(n_clusters = 3)
labels = kmean.fit_predict(X_scaled)
sns.scatterplot(x = X_scaled[:,0], y = X_scaled[:,2], c = labels)
```
